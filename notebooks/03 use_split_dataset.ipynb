# Split Dataset Notebook
Este notebook apoya la división de los conjuntos de datos y cómo usar la clase `RobustDataSplitter` para dividir datasets en train, test y backtest.

## 0. Example

```python
# Cargar datos df
import pandas as pd
parquet_path ="/Workspace/Users/jorgee.lopez@adres.gov.co/mlops_canvas/data/raw/usuarios_unicos.parquet"
df = pd.read_parquet(parquet_path)
print(df.head())
df.shape
```

## 1. Imports

```python
from src.app.split_dataset.robust_data_splitter import RobustDataSplitter
```

## 2. Config DataSplit

```python
# Instanciar el splitter
# Trabajando con series de tiempo entonces usamos el parámetro time
splitter = RobustDataSplitter(
    df,
    split_method='time',
    target_column='usuarios_unicos',
    time_column='mes_registro',
    train_size=0.7,
    test_size=0.2,
    backtest_size=0.1
)
```

```python
# Realizar split
train_df, test_df, backtest_df = splitter.split_data()
# Mostrar resultados
train_df.head()
```

```python
print(f"Train: {len(train_df)} filas")
print(f"Test: {len(test_df)} filas")
print(f"Backtest: {len(backtest_df)} filas")
```

```python
# Calcular métricas de balanceo y estabilidad
metrics = splitter.calculate_metrics()
metrics
```
