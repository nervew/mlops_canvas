# Split Dataset Notebook
Este notebook apoya la división de los conjuntos de datos y cómo usar la clase `RobustDataSplitter` para dividir datasets en train, test y backtest.

## 0. Example

```python
# Cargar datos de ejemplo (Iris)
from sklearn.datasets import load_iris
d = load_iris(as_frame=True)
df = d.frame.copy()
# Asignar columna target
# En este dataset ya existe 'target'
```

## 1. Imports

```python
from src.app.split_dataset.robust_data_splitter import RobustDataSplitter
import pandas as pd
```

## 2. Config DataSplit

```python
# Instanciar el splitter
splitter = RobustDataSplitter(
    df,
    split_method='random',
    target_column='target',
    stratify_columns=['target'],
    train_size=0.7,
    test_size=0.2,
    backtest_size=0.1
)
```

```python
# Realizar split
train_df, test_df, backtest_df = splitter.split_data()
# Mostrar resultados
train_df.head()
```
