```python
# main.py
import yaml
import pandas as pd
from pipeline import FeatureSelectionPipeline

# 1) Carga de configuración
cfg = yaml.safe_load(open("config.yaml"))
method = cfg["feature_selection"]["method"]
params = cfg["feature_selection"]["params"]

# 2) Carga de datos
data = pd.read_csv("datos.csv")
X = data.drop("target", axis=1)
y = data["target"]

# 3) Ejecución de la pipeline
pipeline = FeatureSelectionPipeline(method, params)
X_selected = pipeline.run(X, y)

print("Features seleccionadas:", X_selected.columns.tolist())
```
