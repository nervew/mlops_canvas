```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_parquet("/Workspace/Users/jorgee.lopez@adres.gov.co/mlops_canvas/data/raw/partitioned/train_df.parquet")  
cols = ["EPS_A", "EPS_B", "target"]
sub = df[cols].dropna()
corr = sub.corr()
print("Matriz de correlación:\n", corr)
plt.figure(figsize=(4, 3))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlación EPS_A, EPS_B y Target")
plt.tight_layout()
plt.show()
```
